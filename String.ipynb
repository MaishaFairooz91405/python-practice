```python
#Task 39

s = "abcabcbb"
seen = ""
longest = 0


for i in range(len(s)):

    for j in range(i, len(s)):
        if s[j] in seen:
            break
        seen += s[j]
    longest = max(longest, len(seen))

print(longest)

```
